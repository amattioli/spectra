# Rigel - Beta [Orionis](Index.html)

## Descrizione

Rigel is about as famous as blue supergiants get — it's literally the MK spectral standard that *defines* "B8 Ia" (other stars get classified relative to it, not the reverse), the 7th-brightest star in the sky, and one of the nearest plausible future Type II supernova progenitors known.

| VALD field | Value | Source |
|---|---|---|
| Teff | 12,100 K | Firnstein & Przybilla 2012 |
| log g | 1.75 | Firnstein & Przybilla 2012 |
| Microturbulence | 8 km/s | Firnstein & Przybilla 2012 |
| Composition | held near the "cosmic abundance standard" (≈ solar) for this survey | Firnstein & Przybilla 2012 |
| v sin i (context) | 25 km/s | Firnstein & Przybilla 2012 |

These are well-anchored — the same paper compares against two earlier studies (Takeda & Takada-Hidai, McErlean et al.), and log g = 1.75 is essentially unanimous across all three; Teff shows the only real spread (12,100–13,000 K), which the authors attribute to earlier work not properly accounting for metal line-blanketing.

Same caveat that applies to every luminous supergiant: Hα shows real, variable P-Cygni-type wind signatures, and Rigel also pulsates non-radially with several periods between 1.2 and 74.7 days, on top of longer-term (~100-day) rotationally modulated wind structure from what look like multiple active regions. A static LTE line list handles the bulk of the metal absorption spectrum fine, but won't capture any of that Balmer-line variability.

Rigel is the primary of a wider multiple system — a spectroscopic binary "Rigel B" (itself two stars, Ba+Bb) sits about 9.5″ away, contributing roughly 1/440th of the total light. Negligible at normal spectroscopic slit widths, but worth knowing it's there.
