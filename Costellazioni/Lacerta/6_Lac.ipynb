# 6 [Lacertae](Index.html)

## Descrizione

6 Lac is HD 213420 / HR 8579 / HIP 111104, a young (15.7 Myr) B2 IV subgiant, 12.5 M☉, still fairly early in its post-main-sequence expansion. It's a single-lined spectroscopic binary on an 880-day, moderately eccentric orbit — only the primary's lines show up in the spectrum, so the companion is essentially uncharacterized.

Straightforward case overall: near-solar metallicity, moderate rotation, no peculiarity flags. The only real gap is microturbulence — same recurring story, no dedicated line-by-line analysis turned up for this specific star, so that's an estimate rather than a measurement.
