# Celaeno - 16 [Tauri](Index.html)

## Descrizione

Celaeno is 16 Tauri (HD 23288), one of the Seven Sisters and, like most of the named Pleiades, an extremely fast rotator — 185–212 km/s depending on the source (the two disagree by more than their quoted precision suggests, so treat that as roughly ±15% uncertain). It's a B7 star sitting right at the V/IV classification boundary.

| VALD field | Value | Source |
|---|---|---|
| Teff | 12,800 K | consistent across sources |
| log g | 3.9 | modern catalog |
| [Fe/H] | −0.05 (near solar) | modern catalog |
| Microturbulence | not published — ~1–2 km/s is a reasonable estimate given how hot and fast it spins | — |
| v sin i (context) | ~185–212 km/s (sources disagree) | catalog |

No dedicated line-by-line abundance analysis turned up, so composition and microturbulence are estimates, not measurements. With [Fe/H] close to solar and no peculiarity flag anywhere, solar composition is a safe default.
