# Kappa 2 [Bootis](Index.html)

## Descrizione

κ² Boötis is HD 124675 / HR 5329 / HIP 69483, the brighter, more evolved star of the κ Boo pair (κ¹ Boo, HD 124674, is the fainter F-type companion). This one was in the same Fossati et al. 2008 paper as ι Boo, so there's a matching self-consistent set again:

| VALD field | Fossati et al. 2008 (spectroscopic, VALD-based) | Newer catalog |
|---|---|---|
| Teff | 7500 K | 7,760 K |
| log g | 4.00 | 3.66 |
| Microturbulence | 2.7 km/s | not reported |
| Composition | Fe about −0.26 dex below their solar reference; S enhanced, Cr and Ni depleted, rest close to solar | [Fe/H] = −0.29 dex |
| (context, not a VALD field) v sin i | 121 km/s | 128 km/s |

A few notes specific to this star:

- **Metallicity actually agrees well this time** — Fossati's Fe-line offset (−0.26 dex) and the catalog [Fe/H] (−0.29 dex) are close, so κ² Boo is genuinely a bit metal-poor, not just an artifact of one method. That's a nicer cross-check than ι Boo gave you.
- **Teff and log g disagree more than for ι Boo, and in a consistent direction:** the catalog value is hotter *and* less gravity-bound (log g 3.66 vs 4.00). That tracks with κ² Boo's classification as a slightly evolved A8 subgiant — a subgiant reading (lower log g, from radius+mass) versus an unevolved-dwarf-style spectroscopic fit (higher log g) is exactly the kind of split you'd expect when a star has actually started leaving the main sequence. If your work is sensitive to log g (ionization balance, wing-sensitive species), it's worth considering the lower value seriously rather than defaulting to Fossati's.
- **Abundance pattern is different from ι Boo's** — here it's sulfur enhanced, chromium and nickel depleted, with everything else close to solar, and the paper notes this star has the lowest overall abundance level in their sample. So if you want a custom (not just solar-scaled) composition, those are the three elements to hand-adjust.
- Rotation (~121–128 km/s) is a hair slower than ι Boo's but still fast enough that a denser detection threshold will serve you better than a sparse one.
