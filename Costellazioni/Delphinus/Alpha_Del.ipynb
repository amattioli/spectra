# Alpha [Delphini](Index.html)

## Descrizione

Alpha Del is Sualocin — HD 196867 / HR 7906 / HIP 101958 — and it's a rich: a hierarchical triple where the primary (Alpha Del A / Aa, "Sualocin" itself) is a B9 IV subgiant, orbited every ~17 years by a close pair of an A-type and F-type star (Ba+Bb, on their own 30-day inner orbit). Notably, Aa rotates at almost exactly the same rate as Iota Boo A: v sin i = 144 km/s.

| Teff | 11,643 K | Gardner et al. 2021 (CHARA/MIRC-X interferometry + orbit) |
| log g | 3.93 | Gardner et al. 2021 |
| Microturbulence | not directly measured — see note | — |

Teff/log g here come from a genuinely strong method — dynamical mass, interferometric radius, and SED fitting together, not a photometric calibration. But I couldn't find a dedicated spectroscopic abundance/microturbulence study: interferometric-orbit papers don't do curve-of-growth work, and a 144 km/s rotator is exactly the kind of star line-by-line abundance analyses tend to avoid — the same reason Iota Boo A, at that identical rotation rate, only got tackled by one group willing to do the disentangling.

For microturbulence: hotter B9-type atmospheres trend toward *lower* ξ than the A-type stars earlier in this thread — the decline from ~4 km/s (late A) toward well under 1 km/s (early A/late B) puts a star this hot near the bottom of that scale. I'd start at ξ ≈ 1 km/s absent a better number.

On composition: one online source claims [Fe/H] ≈ −0.06, but that same page had an internally contradictory Teff elsewhere. The fast rotation is actually a useful physical argument here, though — Am/Ap-type peculiarities need a slow, stable atmosphere for diffusion to sort elements, and at 144 km/s that process can't operate, so solar composition is a reasonable default rather than just a placeholder.
