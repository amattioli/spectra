# Beta [Delphini](Index.html)

## Descrizione

Beta Del is Rotanev — HD 196524 / HR 7882 / HIP 101769 — the brightest star in Delphinus, a quite "normal" one: no wind complications, no wild source disagreements, just a straightforward close binary. It's two F5-type stars (giant + subgiant) in a 26.7-year visual orbit, separated by 0.44″.
