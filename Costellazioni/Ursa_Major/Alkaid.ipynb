# Alkaid - Eta [Ursae Majoris](Index.html)

## Descrizione

Alkaid is η Ursae Majoris, the end of the Big Dipper's handle. It is a B3 V main-sequence star that has served as a spectral standard for that type, the same anchor role Rigel and Taygeta play for their types. A very recent (2026) dedicated study found it to be young and fast-spinning: about 5.07 M☉, rotating at roughly 0.41 of critical speed, and only 2–8 Myr past the zero-age main sequence.

**Gravity darkening.** The star is flattened, 2.89 R☉ at the equator vs 2.67 R☉ at the poles, and the model temperature runs from 16,300 K at the equator to 17,700 K at the poles. A single Teff is therefore a disk average, the same caveat as Delta Persei and Atlas's primary, though milder here. The ~16,800 K value sits in the middle of that range.

**Rotation.** Wikipedia's infobox lists the rotation period as 14.6 days, but the paper gives 14.6 hours, which is the only value consistent with a 154 km/s v sin i. An independent photometric power-spectrum peak at about 14.1 hours agrees with it.

**Possible weak disk.** An older spectroscopic study found slightly distorted, variable absorption lines, possibly from weak emission from a disk produced by the rapid rotation. This is a mild Be-like caution for Hα, nowhere near Pleione's level, and the metal lines should be fine. One catalogue entry also carries a spectroscopic-binary ("SB") flag that I couldn't confirm.
