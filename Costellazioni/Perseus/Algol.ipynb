# Algol - Beta [Persei](Index.html)

## Descrizione

Algol is about as famous as stars get — the prototype eclipsing binary, known since antiquity as the "Demon Star," and the textbook example of the **Algol paradox**: the less massive component is the *more* evolved one, because it used to be the heavier star and stripped most of its own mass onto its companion via Roche-lobe overflow. Conveniently, one paper (Kolbas et al. 2015) spectroscopically disentangled all three components and obtained:

| VALD field | Algol A | Algol B | Algol C |
|---|---|---|---|
| Teff | 12,550 K | 4,900 K | 7,550 K |
| log g | 4.05 | 3.25 | 4.18 |
| Microturbulence | 0.4 km/s | not determined | 1.68 km/s |
| [M/H] | −0.03 | not determined | +0.04 |
| v sin i (context) | 50.5 km/s | 62 km/s | 12.4 km/s |

A is the B8V mass-gainer, B is the stripped K-subgiant still filling its Roche lobe (this is the paradox star), C is the more distant A7-type tertiary on a 1.86-year outer orbit. Even this dedicated disentangling effort couldn't pin down a microturbulence or metallicity for B — it contributes under 1% of the system's light in blue, too faint to analyze in detail even after separating the three spectra.

The best part scientifically: Algol A shows the mass-transfer history directly in its own photosphere. Its carbon-to-nitrogen ratio, (C/N)_A = 2.0±0.3, is half the solar value (4.0±0.7) — carbon slightly depleted, nitrogen slightly enhanced, exactly the fingerprint of CNO-processed material that used to be deep inside Algol B and is now sitting on the surface of the star that stole it. Not inferred from theory — measured, in the same spectrum you'd pull lines from.

One correction to the record while I'm here: Algol C gets labeled an "Am star" in a lot of older sources, but this paper's actual element-by-element analysis found it does *not* show the classic Am signature (no strong Sc underabundance, no iron-peak overabundance) — it looks like an ordinary solar-composition A star. Worth knowing before you build it a peculiar composition the way we did for Delta Del.
