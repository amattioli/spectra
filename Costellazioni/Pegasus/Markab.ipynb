# Markab - Alpha [Pegasi](Index.html)

## Descrizione

Markab is HD 218045 / HR 8781 / HIP 113963 — one of the four corners of the Great Square of Pegasus. It's a fast-rotating late-B/early-A star whose evolutionary state is genuinely unsettled in the literature (classified anywhere from B9V dwarf to B9.5III giant to A0IV subgiant depending on the source), which is worth knowing since it means log g carries real ambiguity, not just measurement noise.

| VALD field | Value | Source |
|---|---|---|
| Teff | 8935 K | modern catalog (Wikipedia/Gaia-era) |
| log g | 3.98 (±0.25 — the spread reflects the V/III/IV disagreement) | same |
| Microturbulence | not published — ~1–2 km/s is a reasonable estimate | — |
| [Fe/H] | mildly subsolar to solar; −0.02 (photometric calibration study), −0.10 (a 2003 NLTE spectroscopic Fe study), −0.28 (modern catalog) | three separate sources |
| v sin i (context) | ~130 km/s (125–149 km/s reported across different studies) | various |

A few notes:
- The rotation is fast enough that a precision abundance analysis with a dedicated microturbulence doesn't seem to exist for this one either. Recurring pattern at this point.
- With Teff ~8900 K solidly in "hot, low-microturbulence" B9/A0 territory, I'd lean toward the low end of the estimate (~1 km/s) rather than the ~2–3 km/s appropriate for a cooler A star.
- Every metallicity estimate lands close to solar regardless of method, so solar composition is a safe default — no peculiarity signal here.
