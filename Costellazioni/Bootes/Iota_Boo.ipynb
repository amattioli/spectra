# Iota [Bootis](Index.html)

## Descrizione

For ι Boo itself — the primary, ι Boo A (HR 5350 / HD 125161, the A7V δ Scuti pulsator), not its wide K-dwarf companion HD 234121 — here's what's published, including one paper that did this exact VALD extraction for this exact star:

| VALD field | Fossati et al. 2008 (spectroscopic, used VALD directly on this star) | Newer catalog value |
|---|---|---|
| Teff | 7700 K | 7,764 K |
| log g | 4.45 | 4.14 |
| Microturbulence | 2.6 km/s | not reported |
| Composition | essentially solar Fe (about +0.04 dex above their solar reference), with S, Ti, Y, Ba mildly enhanced | [Fe/H] = +0.19 |
| (context only, not a VALD field) v sin i | 131 km/s | 144 km/s |

I'd lead with the Fossati column — it's a complete, self-consistent set (including the microturbulence, which the catalog entry doesn't give you at all), and it's literally the values used with VALD line data and the LLMODELS/Synth3 pipeline to derive this star's own abundances.

A couple of things worth knowing before you commit to numbers:

- **The 0.3 dex gap in log g (4.45 vs 4.14) is expected, not a red flag.** Iota Boo A is a fast rotator, and Fossati et al. note that spectroscopic log g gets substantially less certain as rotation increases — for a ~130 km/s rotator like this one, they estimate roughly ±0.22 dex, which mostly closes the gap between the two values.
- **Composition is genuinely a bit fuzzy for this star.** Only sulfur, titanium, yttrium, and barium came out overabundant in the detailed analysis; carbon, magnesium, silicon, calcium, scandium, chromium, iron, and nickel all matched solar. So "solar" (or solar +0.1–0.2) is a defensible composition for a general-purpose line list; if you want it tailored to this star specifically, you could hand-enter those four elements and leave the rest at solar defaults — I can help build that abundance block if you want it.
- **v sin i ≈ 130–144 km/s isn't something you enter into Extract Stellar** (it doesn't take rotation as input), but it matters for what you do with the output: at that rotation most of the "significant" lines VALD flags will be blended together in your actual spectrum, so a denser threshold is often more useful here than for a slow rotator.

Wavelength range and detection threshold are really up to your use case rather than something to "look up" — Fossati's team worked across 3872–6943 Å; if you're not trying to match their study specifically, 0.01 is a common threshold for the visibly significant lines, or 0.001 if you want a denser list that captures likely blends.
