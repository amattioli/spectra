# 2 [Lacertae](Index.html)

## Descrizione

2 Lac is HR 8523 — a tight, tidally-distorted double-lined spectroscopic binary: two B6-type stars locked in a 2.617-day circular orbit, too close to resolve visually (you only know it's two stars from the doubled, Doppler-shifted spectral lines). Nice contrast to Algol: there, the *less* massive star was the more evolved one because of mass transfer. Here it's the ordinary case — the more massive component (6.99 M☉) is also the more evolved one, sitting near the very end of its main-sequence lifetime (over 90% of the way through it), while the less massive secondary (5.55 M☉) is still comfortably unevolved. Same birth, same age, different pace — mass sets the clock.

| VALD field | Primary | Secondary |
|---|---|---|
| Teff | 15,000 K | close to the primary's (Hilditch's classical analysis found them similar in Teff despite the age gap) |
| log g | 3.0 | not directly published — likely ~4.0–4.2, a normal ZAMS B6V value, since it's essentially unevolved |
| Microturbulence | not published | not published |
| Composition | assumed solar | assumed solar |
| v sin i (context) | 50 km/s | not confirmed — tidally synchronized, so scales with its (smaller) radius |

Ending up at nearly the same Teff despite such different log g is itself the interesting bit: it's what evolutionary tracks predict for two coeval stars at these masses — the primary has ballooned in radius as it exhausts core hydrogen (dropping log g while Teff holds roughly steady), while the secondary hasn't started that expansion yet.

No dedicated abundance work turned up for this system, so solar composition is the reasonable default for both.
