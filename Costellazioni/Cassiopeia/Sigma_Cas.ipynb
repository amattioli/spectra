# Sigma [Cassiopeiae](Index.html)

## Descrizione

Sigma Cas is HD 224572 / HR 9071 / HIP 118243 (8 Cas), is a member of the hot B-pulsator family like Mirzam, Furud, Adhara. The primary is a B2 IV subgiant with a B3 V companion about three arcseconds away. An older catalogue instead classifies it as B1 V. That matters for log g: a B1 V would sit nearer 4.0, while the 3.6 below fits the evolved reading.

**Pulsation status.** A dedicated spectroscopic study by Catanzaro and colleagues placed it inside the β Cephei instability region and found line-profile variability with excess power at 2–8 cycles per day, peaking at 3.00 and 5.32 cycles per day. That is the right range for β Cep pulsation, but it's still a candidate, not a settled pulsator like Mirzam. In well-studied β Cep stars the instantaneous Teff and log g swing by roughly 700 K and 0.2 dex over the cycle. So treat the 21,000 K / 3.6 pair as a time average, which is fine for a line list.

**Microturbulence gap.** That Catanzaro paper also derived atmospheric parameters and abundances, but I couldn't open it (the site blocked access). So my ξ is an analogy, not a measurement for this star. β Cep itself came out at 8.1 ± 0.9 km/s in the same group's work, and other early-B pulsators have given 6±3 and 10±4 km/s. For VALD's depth threshold the exact value matters only modestly.

**Companion.** B is about 2.1 magnitudes fainter and 3.1″ away.
