# Maia - 20 [Tauri](Index.html)

## Descrizione

Maia is genuinely the odd one out among the named Pleiades — while Celaeno, Atlas, and the rest spin at 130–220 km/s, Maia rotates at just 33 km/s, the slowest bright B star in the cluster by a wide margin. That slowness is the whole story here: it's what let heavy-element diffusion actually take hold in its atmosphere, making Maia a confirmed **HgMn (mercury-manganese) star** — a different chemically peculiar class distinct from the Am stars (Sirius, Delta Del) and the magnetic Ap/He-weak class (Atlas's secondary).

Two things worth knowing:

**The Fe overabundance is real, not a data error.** Unlike some suspicious single-source numbers, +1.10 dex is actually consistent with how HgMn stars behave — the signature is wild, element-specific overabundances (manganese and mercury especially, sometimes enhanced by factors of 100,000×), and iron itself commonly comes out enhanced by close to a dex in this class. If you want a composition that's genuinely representative rather than a flat +1.10 scaling, Mn and Hg need far larger individual boosts than Fe — I can look up typical HgMn enhancement factors for those specifically if that's useful.

**Best story arguably.** Maia is the literal namesake of the "Maia variable" class — proposed in the 1950s by Otto Struve as a category of rapidly pulsating hot stars, with Maia as the prototype. Every attempt to confirm pulsation in Maia itself failed for sixty years, to the point where researchers doubted the class was real at all. Then in 2016, Kepler/K2 photometry finally detected genuine brightness variability — a clean 10.3-day signal. It turned out to come from a chemical spot (the same kind of overabundant-element patch behind the HgMn classification) rotating in and out of view, not pulsation at all. The paper that found this makes the point outright: Maia is variable, but Maia is not a Maia variable.
