# Cursa - Beta [Eridani](Index.html)

## Descrizione

Cursa is Beta Eridani — HD 33111 / HR 1666 / HIP 23875 — a rapidly rotating, evolved A3 giant/subgiant (sources disagree on III vs IV), and notably one of a small handful of stars known to produce enormous sudden flares: it flared in 1985, brightening by three magnitudes for over two hours, a phenomenon shared by only about a dozen known stars (Enif and Mu Cephei among them).

| VALD field | Value | Source |
|---|---|---|
| Teff | 8360 K (8357±237) | consistent across essentially every source |
| log g | 3.58 | modern catalog |
| [Fe/H] | +0.03 (essentially solar) | modern catalog |
| Microturbulence | not published — ~2–3 km/s is a reasonable estimate for an evolved A-type at this log g | — |
| v sin i (context) | 196 km/s | consistent across every source — one of the fastest rotators in this whole thread |

Worth flagging: mass, radius, and luminosity are genuinely inconsistent across sources (mass 2.0–2.55 M☉, radius 2.4–3.14 R☉, luminosity 25–45 L☉ depending on the catalog), even though Teff/log g/[Fe/H] agree well. That's most likely different pipelines making different assumptions from the same underlying photometry, not a real physical ambiguity — but if precise mass/radius matters for your work, it's worth tracing which specific study a given number comes from.

Composition is solar with high confidence.
