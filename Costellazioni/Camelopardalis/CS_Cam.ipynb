# CS [Camelopardalis](Index.html)

## Descrizione

CS Cam is HD 21291 / HR 1035 — a near-twin sibling to CE Cam: another α Cygni-variable blue supergiant in the same Cam OB1 association (VdB 14 nebula, right next to CE Cam's VdB 15), but hotter and slightly less evolved: **B9 Ia** rather than A0 Iae. It's also a wide binary — a fainter B2 III companion (V=7.75) sits alongside it, though at a somewhat different estimated distance, so it's likely just a wide/optical pair rather than a bound system.

CS Cam *is* one of the 35 stars in Firnstein & Przybilla (2012), the homogeneous non-LTE BA-supergiant survey — so we have a complete, directly-measured set:

| VALD field | Value | Source |
|---|---|---|
| Teff | 10800 K | Firnstein & Przybilla 2012 |
| log g | 1.65 | Firnstein & Przybilla 2012 |
| Microturbulence | 7 km/s | Firnstein & Przybilla 2012 |
| Composition | held near the young-Population-I "cosmic abundance standard" (≈ solar) in that survey | Firnstein & Przybilla 2012 |
| v sin i (context) | ~30–32 km/s | Wikipedia catalog / Firnstein & Przybilla |

A couple of notes:
- Same caveat as CE Cam applies to Hα (and possibly Hβ/Hγ, given the high luminosity): this survey's own methodology excludes wind-affected Balmer lines from quantitative fits, so expect the lower Balmer series to look different from what a static LTE line list predicts. Metal lines should behave fine.
- Rotation (~30 km/s) is close to CE Cam's, so still a comparatively clean, slowly-rotating spectrum.
