# Alpha [Lacertae](Index.html)

## Descrizione

Alpha Lac is "Stellio" — HD 213558 / HR 8585 / HIP 111169 — and it's a genuine single star (the visual companion 36″ away is a confirmed chance alignment, not physically bound), an ordinary A1V dwarf still on the main sequence.

| VALD field | Value | Source |
|---|---|---|
| Teff | 9050 K | consistent across catalog sources |
| log g | ~4.15 (computed from mass/radius — no direct spectroscopic value published) | derived |
| Microturbulence | not published — ~1–2 km/s is a reasonable estimate given how hot and fast it rotates | — |
| Composition | assumed solar (no dedicated abundance study found) | — |
| v sin i (context) | 128 km/s | catalog |

At 128 km/s rotation, this isn't the kind of star that attracts detailed line-by-line abundance work, so there's no published [Fe/H] or microturbulence that I could find. The log g is my own calculation from the published mass (2.19 M☉) and radius (2.14 R☉) via g = GM/R² and not a direct spectroscopic measurement.

Nothing here points to chemical peculiarity — plain A1V classification, no Am/Ap flags anywhere — so solar composition is a safe, uncomplicated default this time.
