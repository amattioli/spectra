# Taygeta - 19 [Tauri](Index.html)

## Descrizione

Taygeta is 19 Tauri Aa — the officially-named component of a triple-ish system: a genuine close spectroscopic companion (Ab, 1313-day orbit) plus an unrelated, much wider "B" star (69″ away) that just happens to share the name but isn't physically bound. It's also an MK spectral standard — the reference star that defines B6IV, same status as Rigel has for B8Ia.

| VALD field | Taygeta (Aa) | Ab (companion) |
|---|---|---|
| Teff | 13,696 K | 8,306 K |
| log g | 3.77 | 4.2 |
| Microturbulence | not published | not published |
| Composition | assumed solar (no dedicated study found) | assumed solar |
| v sin i (context) | 105 km/s | — |

No dedicated abundance or microturbulence analysis turned up for this one, and its MK-standard status apparently hasn't been enough to attract detailed line-by-line work.

The genuinely fun fact about this star is how *un*-interesting it is photometrically, which stands out after Pleione and Maia: it was once flagged as a suspected variable, but SIMBAD has since formally retracted that — explicitly noting the claim couldn't be confirmed — and separate studies found it among the least variable stars in the entire Hipparcos catalogue. Also worth knowing if you're trying to match an observed spectrum: the companion took decades to nail down, with repeated detections and non-detections across different methods (radial velocity, lunar occultation, speckle interferometry) before it was unambiguously confirmed in 1994 — so older literature on this star may disagree about whether it's even a binary.
