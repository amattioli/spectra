# Rigel - Beta [Orionis](Index.html)

## Descrizione

Saiph is κ Orionis (HD 38771), the "other knee" of Orion opposite Rigel — much hotter (B0.5 Ia vs B8 Ia) even though it looks fainter, since most of its light comes out in the UV where our eyes can't see it. There's a dedicated, well-cited non-LTE study covering exactly this star (Crowther, Lennon & Walborn 2006), which hands you a complete parameter set:

| VALD field | Value | Source |
|---|---|---|
| Teff | 26,500 K | Crowther, Lennon & Walborn 2006 |
| log g | 2.9 (true) / 2.7 ("effective," radiation-pressure-corrected) | same |
| Turbulence | 12.5 km/s | same |
| Composition | held at solar (Grevesse & Sauval 1998), individual CNO derived separately | same |
| v sin i (context) | 83 km/s | same — matches an independent catalog value exactly |

Use 2.9 for VALD, not 2.7 — "effective gravity" is a bookkeeping device for how close a star sits to its Eddington limit (subtracting the radiative acceleration from electron scattering), not the actual surface gravity a hydrostatic model atmosphere needs.

Two things worth flagging:

**That 12.5 km/s "turbulence" isn't the same animal as classical microturbulence in cooler stars.** At this temperature the authors explicitly treat it as *radially varying*, starting from a canonical 20 km/s guess and adjusting between 10–40 km/s per star to fit the helium and silicon lines — it's standing in for shock-driven turbulence from wind instabilities, not simple photospheric thermal motion. Saiph's 12.5 km/s sits on the low side of that range.

**Distance to this star is genuinely disputed** — echoes the Pleiades/Atlas situation. Hipparcos gives ~220 pc, but the authors instead adopt ~400 pc based on Saiph's membership in the Ori OB1c subgroup, arguing individual Hipparcos parallaxes aren't reliable at this distance and reddening. That choice affects luminosity and radius, not Teff/log g/turbulence.

This same paper ran a clumped-wind test specifically on HD 38771 (10% volume filling factor) to try to explain why models over-predict a UV Si IV wind feature — it didn't fully work, and the puzzle's still open.
