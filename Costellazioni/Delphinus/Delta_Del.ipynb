# Delta [Delphini](Index.html)

## Descrizione

Delta Delphini isn't just *a* peculiar star — it's the prototype and namesake of an entire class, the "δ Delphini stars," characterized in Don Kurtz's landmark 1976 study. A few reasons it earned that status:

- It was the 5th δ Scuti variable ever recognized (Eggen, shortly after the original four in 1956).
- Its classification, **kA7hF1VmF1pSrEuCr:**, is a textbook metallic-line (Am) signature: the Ca II K-line reads like A7, but the hydrogen and metal lines read like F1, with strontium, europium, and chromium conspicuously overabundant — the classic fingerprint of gravitational settling sorting elements in a calm atmosphere.
- That's normally a contradiction with pulsating: Am-star diffusion needs a still atmosphere, but δ Scuti driving needs helium sitting in the ionization zone, which diffusion strips out — so classical *main-sequence* Am stars essentially don't pulsate. δ Del does anyway, because it's evolved off the main sequence, which is exactly the loophole Kurtz explored: pulsation and metallicism, mutually exclusive early on, can coexist once a star evolves.
- Both stars in this 40.6-day double-lined binary show essentially the same peculiar classification *and* both pulsate — not a one-off fluke.
- Bonus oddity: the modern interferometric+RV orbit (Gardner et al. 2018) found the two components' masses, radii, and flux ratio don't fit a common age under standard evolution models — a ~200 Myr mismatch hinting at a non-standard history, possibly mass transfer involving a now-vanished third star.

On composition — this is one case where I'd push back on using a single "[Fe/H]" at all. Sources disagree (−0.5 in the modern catalog vs. an unreliable −0.2 elsewhere), but more to the point, a bulk metallicity misses what makes this star interesting: it's not uniformly metal-rich or -poor, it's *patterned* — Ca and Sc reading artificially depleted, Sr/Eu/Cr genuinely enhanced, roughly solar elsewhere. A VALD list that's actually representative of δ Del would come from hand-adjusting those specific elements, not from any single scaled composition. Kurtz (1976) did exactly this fine abundance work for δ Del itself against standard comparison stars — I can try to dig up his actual derived values (old paper, may hit the same paywall trouble as CE Cam) if you want real numbers instead of "Am-typical" enhancements.

For microturbulence, with no direct measurement surfacing: F-subgiants in this Teff/log g range typically run ξ ≈ 2–3 km/s, a reasonable starting point for both stars absent a better source.
