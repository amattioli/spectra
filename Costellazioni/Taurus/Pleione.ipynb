# Pleione - 28 [Tauri](Index.html)

## Descrizione

Pleione (28 Tau, BU Tauri, HD 23862) is one of the most famous Be/shell stars there is.

| VALD field | Value | Note |
|---|---|---|
| Teff | ~12,000 K | B8IVpe/B8Vne — sources disagree on luminosity class |
| log g | ~3.96 (computed from mass/radius) | not directly published |
| Microturbulence | not published | no dedicated abundance study — same recurring gap |
| Composition | assumed solar | no dedicated abundance study found |
| Rotation | near-critical (close to breakup) | this *is* the defining Be mechanism, not incidental context this time |

Here's the thing that matters more than any of those numbers: Pleione doesn't have one spectrum. It cycles between three genuinely different appearances — normal B star, Be (emission), and Be-shell (narrow absorption superimposed on emission) — on a roughly 34.5-year cycle, as a circumstellar decretion disk forms, precesses, and dissipates. It spent 1905–1936 as a plain B star, then alternated Be/shell phases through the 20th century, went into a shell phase around 2006, and — as of the most recent reports I can find (mid-to-late 2025) — has swung back to being viewed nearly pole-on (~60° disk inclination) again, meaning it's likely back to a Be-type appearance rather than shell. There was also a genuinely unique episode in 2005–2007 where it briefly carried *two* misaligned disks at once (~30° and ~60° inclination) — the first double-disk system ever observed around any star.

There's also an unresolved multiplicity picture worth knowing about: a close companion at 0.24″, a wider M5 companion at 4.66″, and a proposed ~218-day spectroscopic companion inferred from Hα radial-velocity variations — the last one is plausibly what drives the disk precession, but its orbit (and even its reality) is still debated in the literature.
