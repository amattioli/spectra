# Atlas - 27 [Tauri](Index.html)

## Descrizione

Atlas (27 Tau, HD 23850) is the second-brightest star in the Pleiades, and it's got real history: it was one of the key binaries used to settle a genuine controversy — Hipparcos said the Pleiades sit at ~118 pc, ground-based methods insisted on ~130 pc, and precise orbits like this one (now 136.2±1.4 pc) proved Hipparcos was wrong. 

| VALD field | Primary | Secondary |
|---|---|---|
| Teff | 12,525 K | 12,835 K |
| log g | 3.38 | 4.20 |
| Microturbulence | 0.5 km/s | 1.0 km/s |
| [M/H] | −0.03 (solar) | 0.0, fixed (see note) |
| v sin i (context) | 217 km/s | 47 km/s |

Two things make this system special:

**The primary is directly, interferometrically confirmed to be egg-shaped.** At v sin i = 217 km/s, it's spinning fast enough that CHARA long-baseline interferometry actually resolved the distortion — polar-to-equatorial radius ratio ≈ 0.83. Only about a dozen stars have ever had their shape measured this way, and Atlas is by far the most distant of them. Models predict the pole runs roughly 650 K hotter than the equator — so, same caveat as Delta Persei a few turns back, but here it's *measured* rather than inferred: a single Teff/log g is necessarily a disk-averaged approximation.

**The secondary is a newly-confirmed magnetic, chemically peculiar (ApBp) star** — and this 2025 paper is what actually nailed down *which* of the two stars carries the peculiarity. It's been known since 2009 that the Atlas system was "helium-weak," and since 2015 that it carries a several-hundred-Gauss magnetic field, but nobody knew which component until now. It's the secondary: helium runs at about 1/8 solar, while iron, chromium, and titanium are all enhanced roughly tenfold. If you want a real custom composition instead of flat solar:

| Element | Secondary, relative to solar |
|---|---|
| He | −0.90 dex |
| Fe | +1.05 dex |
| Cr | +0.95 dex |
| Ti | +0.80 dex |

Everything else stays solar for the secondary; the primary's composition is ordinary solar throughout — the peculiarity belongs to the secondary alone.

One flag: the paper notes the secondary shows time-variable line-profile distortions — rotating chemical spots, essentially — so any single composition is a time-average of a star with real patchy surface abundances.
