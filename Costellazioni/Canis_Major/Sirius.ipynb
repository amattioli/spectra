# Sirius - Alpha [Canis Majoris](Index.html)

## Descrizione

Sirius A (HD 48915) is one of the most intensively studied stellar atmospheres that exists, and there's a landmark paper (Landstreet 2011) that nails down every parameter.

| VALD field | Value | Source |
|---|---|---|
| Teff | 9900 K | Landstreet 2011 — Strömgren colours and luminosity+radius agree |
| log g | 4.30 | same — from a very precisely known mass (2.12 M☉) and interferometric radius |
| Microturbulence | 2.2 km/s | same — independently verified at two different atmospheric depths (above and below the Balmer jump), found consistent |
| Composition | confirmed hot Am star — see pattern below | same |
| v sin i (context) | 16.5 km/s | Landstreet 1998 |

That last line matters more than it looks: Sirius A is a genuine slow rotator, which is *why* it's such a clean, tractable Am star in the first place.

**On composition**, this isn't a case for flat solar scaling — Sirius A has one of the most completely measured abundance patterns of any A star (20 of 27 elements from He to Ni individually determined, plus Sr/Y/Zr/Ba). The broad strokes: carbon depleted by roughly 1 dex, boron depleted by a factor of 30+, scandium and calcium depleted, while the iron-peak elements (Cr, Mn, Co, Ni) and heavy elements (Sr, Y, Zr, Ba) are enhanced by factors of 3–30 — the classic Am diffusion signature, but unusually well quantified. 

Like Algol, there's a mass-transfer hypothesis here too — Sirius B used to be the more massive star (~6 M☉) and evolved into a white dwarf first, and this paper argues Sirius A may have accreted up to 0.5 M☉ of AGB-processed material from it, which would explain the odd C/N/heavy-element pattern better than simple diffusion alone does.
