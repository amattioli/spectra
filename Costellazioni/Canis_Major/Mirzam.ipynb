# Mirzam - Beta [Canis Major](Index.html)

## Descrizione

Mirzam is genuinely tied up in the naming history of an entire pulsation class: "Beta Cephei variables" are also literally called "Beta Canis Majoris stars" in the older literature, because β CMa was one of the first stars recognized to pulsate this way — it's a real rivalry for prototype status, even though β Cephei itself won out as the more commonly used name. Mirzam itself pulsates in (at least) three simultaneous close frequencies, giving it the classic beat-phenomenon light and velocity curves that define the class.

| VALD field | Value | Source |
|---|---|---|
| Teff | ~25,000 K (24,700–25,200 K across three independent determinations) | modern catalog + Niemczura & Daszyńska-Daszkiewicz 2004 |
| log g | ~3.70–3.74 | same sources |
| Microturbulence | 2 km/s | Niemczura & Daszyńska-Daszkiewicz 2004 (see note) |
| [Fe/H] | +0.04 (essentially solar) | confirmed independently by two sources |
| v sin i (context) | 31 km/s | see reconciliation note below |

A few notes:

**The microturbulence isn't a per-star fit — it's a fixed assumption.** The one paper I found with an explicit ξ value for β CMa (a UV-spectrophotometric study of the whole β Cephei class) held microturbulence at a standard 2 km/s for every star in their sample rather than deriving it individually. It's a reasonable, literature-standard number for a star this hot, just worth knowing it wasn't measured specifically for Mirzam.

**The v sin i literature actually disagrees by a factor of 3** (31±5 km/s vs. an older 11±7 km/s): Mirzam has an independently measured 13.6-day rotation period, and combined with its radius (8.44 R☉), that implies an equatorial velocity of ~31 km/s — matching the higher value almost exactly, which means we're seeing the star close to equator-on. I'd trust 31 km/s.

**Composition is genuinely solid**, not just assumed — +0.04 dex shows up independently in a modern catalog and in dedicated UV spectrophotometry, so solar composition is a well-supported choice here, not a fallback.
