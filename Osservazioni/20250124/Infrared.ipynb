# Confronto nell'infrarosso tra le osservazioni del [24/01/2025](Index.html)

Il vicino infrarosso è fortemente inquinato dalle bande telluriche dell'acqua e dell'ossigeno molecolare.
Confrontando i tre spettri presi nella stessa serata, di stelle tutte angolarmente molto vicine tra di loro, è ragionevole supporre che le le righe telluriche siano in comune ai tre, mentre le differenze siano dovute a caratteristiche stellari.
Il tentativo di identificare le righe stellari non è però agevole a causa della sovrapposizione, a partire da $\lambda$7500 circa, dello spettro del secondo ordine. Occorrerebbe ripetere le osservazioni usando un filtro IR pass 650 che al momento non possiedo
