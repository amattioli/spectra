# HD 17505 in [Camelopardalis](Index.html)

## Descrizione

HD 17505 is a genuine **O-star multiple system**, not a supergiant. It's at least a hierarchical triple (with several more distant, unconfirmed visual companions): a close 8.57-day binary of two near-twin O7 V((f)) stars (Aa1, Aa2), orbited by a third star, an O6.5 III (Ab), on a >100-year outer orbit. Unqualified "HD 17505" usually means the whole blended system, but the only real component-level parameters come from disentangling Aa1/Aa2 specifically (Raucq et al. 2018, using the NLTE code CMFGEN).

| VALD field | Aa1 (primary) | Aa2 (secondary) |
|---|---|---|
| Teff | 37,000 K | 36,700 K |
| log g | 3.75 | 3.74 |
| Microturbulence | 10 km/s (photospheric anchor of their depth-dependent wind model) | same |
| Composition | near-solar; N mildly enhanced, C/O mildly depleted (rotational mixing signature) | same pattern |

These two are close to interchangeable — same spectral type, Teff within 300 K, log g within 0.01 dex. I'd just pick one, or average them, unless you specifically need the disentangled pair.

One caveat here is: VALD's Extract Stellar selects lines using a static LTE Kurucz/Castelli atmosphere, and a real O7 star has a strong wind plus significant NLTE effects — that's exactly why this paper needed CMFGEN instead of a simple grid lookup. VALD will still accept Teff=37000 K/log g=3.75 and hand you a line list, and it's still a legitimate *source of atomic data* (wavelengths, gf-values, damping constants — the raw ingredients NLTE codes pull from too). But the predicted line depths/significance behind the threshold cut won't be very trustworthy here, especially for wind-sensitive lines (Hα, He II 4686, the C/N/O "f"-classification features) that aren't well described by a static photosphere. Treat the output as "candidate metal lines in this range," not "what the spectrum looks like."

The tertiary (Ab, O6.5 III) wasn't independently fit in this paper — it was held fixed at literature-typical O6.5 III values while disentangling the inner pair — so I don't have a precise Teff/log g for it specifically. Want me to track one down?
